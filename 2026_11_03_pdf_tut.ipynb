# Best for standard text extraction

```python
from pypdf import PdfReader
```

```python
# load the PDF file
reader = PdfReader("Python PDF Test Document.pdf")
```

```python
# get the total number of pages
print(f"Total pages: {len(reader.pages)}")
```

```python
# extract and print text from all pages
for index, page in enumerate(reader.pages):
    text = page.extract_text()
    print(f"--- Page {index + 1} ---")
    print(text)
```

```python
reader = PdfReader("Python PDF Test Document.pdf")
page = reader.pages[0]
text = page.extract_text()
print(text)
```

# Best for preserving visual layouts and tables

```python
import pdfplumber
```

```python
# open the PDF using a context manager
with pdfplumber.open("Python PDF Test Document.pdf") as pdf:
    first_page = pdf.pages[0]
    tables = first_page.extract_tables()
    for row in tables[0]:
        print(row)
```

# Best for speed and performance

```python
import pymupdf
```

```python
# Open the document
doc = pymupdf.open("Python PDF Test Document.pdf")

for page_num in range(len(doc)):
    page = doc.load_page(page_num)
    text = page.get_text()
    print(f"--- Page {page_num + 1} ---")
    print(text)
```

```python
import fitz
```

```python
doc = fitz.open("Python PDF Test Document.pdf")
page = doc[0]
blocks = page.get_text("blocks")
for b in blocks:
    print(b)
```

# exploring pymupdf4LLM

```python
import pathlib
import pymupdf4llm

# Convert PDF to Markdown string
md_text = pymupdf4llm.to_markdown("Python PDF Test Document.pdf")

# Save output as a Markdown file
pathlib.Path("output.md").write_bytes(md_text.encode("utf-8"))
print("Successfully saved PDF as output.md")
```

```python
# Convert PDF to Markdown string
md_text = pymupdf4llm.to_markdown("Python Multi-Page Test Document.pdf")

# Save output as a Markdown file
pathlib.Path("output_multi.md").write_bytes(md_text.encode("utf-8"))
print("Successfully saved PDF as output.md")
```
