# PyMuPDF (fitz) Multi-Page Testing Script

This script executes a comprehensive testing pipeline on multi-page PDF files, demonstrating text extraction, bounding box lookup, table parsing, link extraction, and image rendering.

```python
"""
PyMuPDF Multi-Page Testing Script

This script executes a comprehensive testing pipeline on multi-page PDF files,
demonstrating text extraction, bounding box lookup, table parsing, link extraction,
and image rendering.

Requirements:
pip install pymupdf
"""

import os
import sys

import pymupdf

def test_pymupdf_pipeline(pdf_path: str):
    print("=" * 60)
    print(" PyMuPDF Multi-Page Testing Suite")
    print("=" * 60)

    # 1. Open Document & Extract Metadata
    if not os.path.isfile(pdf_path):
        print(f"[ERROR] File not found: '{pdf_path}'")
        return

    try:
        doc = pymupdf.open(pdf_path)
    except Exception as e:
        print(f"[ERROR] Failed to open PDF file '{pdf_path}': {e}")
        return

    with doc:
        metadata = doc.metadata or {}
        print("\n[1] DOCUMENT OVERVIEW")
        print(f" - File Path:   {pdf_path}")
        print(f" - Total Pages: {len(doc)}")
        print(f" - Title:       {metadata.get('title') or 'N/A'}")
        print(f" - Author:      {metadata.get('author') or 'N/A'}")
        print(f" - Format:      {metadata.get('format') or 'N/A'}")

        # 2. Iterate Pages & Extract Text Summary
        print("\n[2] PAGE-BY-PAGE TEXT & BLOCK SUMMARY")
        for page_num, page in enumerate(doc):
            text = page.get_text("text")
            blocks = page.get_text("blocks")
            words = page.get_text("words")

            print(f"\n  --- Page {page_num + 1} of {len(doc)} ---")
            print(f"  * Dimensions: {page.rect.width:.1f}pt x {page.rect.height:.1f}pt")
            print(f"  * Text Length: {len(text)} characters")
            print(f"  * Text Blocks: {len(blocks)} identified")
            print(f"  * Word Count:  {len(words)} words")

            # Display first line preview
            stripped = text.strip()
            first_line = stripped.split("\n")[0] if stripped else "EMPTY PAGE"
            print(f"  * Header Preview: \"{first_line[:60]}\"")

        # 3. Keyword Search & Coordinate Bounding Boxes
        search_term = "CRITICAL_SEARCH_TARGET_2026"
        print("\n[3] KEYWORD SEARCH & BOUNDING BOX EXTRACTION")
        print(f" Searching for string: '{search_term}'...")

        matches_found = 0
        for page_num, page in enumerate(doc):
            rects = page.search_for(search_term)
            matches_found += len(rects)
            for rect in rects:
                print(f"  [MATCH] Page {page_num + 1} at bounding box "
                      f"(x0={rect.x0:.2f}, y0={rect.y0:.2f}, x1={rect.x1:.2f}, y1={rect.y1:.2f})")

        if matches_found == 0:
            print("  [INFO] Search term not found. Testing fallback query 'Performance'...")
            for page_num, page in enumerate(doc):
                rects = page.search_for("Performance")
                if rects:
                    print(f"  [MATCH] 'Performance' found on Page {page_num + 1} ({len(rects)} occurrences)")

        # 4. Table Extraction (page.find_tables, requires PyMuPDF >= 1.23)
        print("\n[4] TABLE DETECTION & EXTRACTION")
        table_count = 0
        for page_num, page in enumerate(doc):
            try:
                tabs = page.find_tables()
            except AttributeError:
                print("  [WARN] find_tables() not available - upgrade PyMuPDF (>= 1.23).")
                break
            for table in tabs.tables:
                table_count += 1
                rows = table.extract()
                n_cols = len(rows[0]) if rows else 0
                print(f"  * Found Table #{table_count} on Page {page_num + 1} "
                      f"({len(rows)} rows x {n_cols} cols)")
                # Print header and first 2 rows
                for row_idx, row in enumerate(rows[:3]):
                    clean_row = ["" if cell is None else str(cell).replace("\n", " ")[:30] for cell in row]
                    prefix = "Header: " if row_idx == 0 else f"Row #{row_idx}: "
                    print(f"    {prefix}{clean_row}")

                # Print header and all rows
                for row_idx, row in enumerate(rows[:-1]):
                    clean_row = ["" if cell is None else str(cell).replace("\n", " ")[:30] for cell in row]
                    prefix = "Header: " if row_idx == 0 else f"Row #{row_idx}: "
                    print(f"    {prefix}{clean_row}")
                    
        if table_count == 0:
            print("  * No tables detected.")

        # 5. Extract Hyperlinks & Annotations
        print("\n[5] HYPERLINKS & ANNOTATIONS")
        total_links = 0
        for page_num, page in enumerate(doc):
            links = page.get_links()
            total_links += len(links)
            for link in links:
                link_type = link.get("kind")
                target = link.get("uri") or link.get("page")
                print(f"  * Page {page_num + 1} Link: Type={link_type}, Target={target}")
        if total_links == 0:
            print("  * No links found.")

        # 6. Page Rendering to Image (Thumbnail Generation)
        print("\n[6] PAGE RENDERING (THUMBNAIL PREVIEW)")
        for page_num in range(min(2, len(doc))):  # Render first 2 pages
            page = doc[page_num]
            pix = page.get_pixmap(dpi=150)  # Render at 150 DPI
            out_filename = f"page_{page_num + 1}_preview.png"
            pix.save(out_filename)
            print(f"  * Saved Page {page_num + 1} raster preview -> '{out_filename}' "
                  f"({pix.width}x{pix.height} px)")

    print("\n" + "=" * 60)
    print(" PyMuPDF Test Execution Completed Successfully!")
    print("=" * 60)

if __name__ == "__main__":
    # Pick the first argument that looks like a PDF. This ignores extra
    # arguments Jupyter/Colab inject (e.g. "-f kernel.json").
    pdf_args = [a for a in sys.argv[1:] if a.lower().endswith(".pdf")]
    pdf_filename = pdf_args[0] if pdf_args else "Python Multi-Page Test Document.pdf"
    test_pymupdf_pipeline(pdf_filename)
```
