# A Python fundamentals walkthrough 

```python
# create a class
class MyClass:
    x = 5

# create an object of the class MyClass
p1 = MyClass()
print(type(p1))
print(p1.x)
```

```python
# all classes have a built-in method called __init__(), which is always executed when the class is being initiated 
class Person:
    def __init__(self,name,age):         # constructor
        self.name = name
        self.age = age

p1 = Person("Emil", 36)

print(p1.name)
print(p1.age)
```

```python
class Person:
    def __init__(self,name,age=18):       # constructor
        self.name = name
        self.age = age

p1 = Person("Emil")
p2 = Person("Tobias",25)

print(p1.name,p1.age)
print(p2.name,p2.age)
```

```python
# The self parameter links the method to the specific object
class Person:
    def __init__(self,name):
        self.name = name
        
    def printname(self):
        print(self.name)
        
p1 = Person("Tobias")
p2 = Person("Linus")

p1.printname()
p2.printname()
```

```python
# the self can be named as something else as well
class Person:
    def __init__(myobject,name,age):
        myobject.name = name
        myobject.age = age

    def greet(abc):          # myobject and abc point to the same thing
        print("Hello, my name is " + abc.name)

p1 = Person("Emil",36)
p1.greet()
```

```python
# try-except statements
x = "Anonymous"
try:
    print(x)
except NameError:
    print("variable x is not defined!")
except:
    print("something went wrong!")
else:
    print("nothing went wrong!")
finally:
    print("the try-except statement is finished!")
```

```python
# the built-in range() function
x = range(10)
print(x)

x = range(3,10,2)
print(x)

for i in range(10):
    print(i)

print(list(range(5,20,3)))
```

# A lambda function is a small anonymous function
## `lambda arguments : expression`

```python
x = lambda a: a+10
print(x(5))
```

```python
x = lambda a,b: a*b
print(x(5,6))
```

```python
x = lambda a,b,c: a+b+c
print(x(5,6,2))
```

## The `map()` function applies a function to every item in an iterable

```python
numbers = [1,2,3,4,5]
doubled = list(map(lambda x: x*2,numbers))
print(doubled)
```

# `*args` (arguments) and `*kwargs` (keyword arguments)

```python
def my_function(*args):
    print("Type:",type(args))
    print("First argument:", args[0])
    print("Second argument:", args[1])
    print("All arguments:", args)

my_function("Emil","Tobias","Linus")
```
