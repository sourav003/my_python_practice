```python
# convert from JSON to Python
import json

# some random json
x = '{"name":"John", "age":30, "city":"New York"}'

# parse x:
y = json.loads(x)

# the result is a Python dictionary:
print(y["age"])
```

```python
# convert from Python to JSON

# a Python object (dict)
x = {
    "name": "John",
    "age": 30,
    "city": "New York"
}

# convert into JSON
y = json.dumps(x)

# the result is a JSON string
print(y)
```

### Any Python object can be converted to JSON

```python
import json

print(json.dumps({"name": "John", "age": 30}))
print(json.dumps(["apple", "bananas"]))
print(json.dumps(("apple", "bananas")))
print(json.dumps("hello"))
print(json.dumps(42))
print(json.dumps(31.76))
print(json.dumps(True))
print(json.dumps(False))
print(json.dumps(None))
```

```python
# convert a Python object containing all the legal data types to JSON 
import json

x = {
    "name": "John",
    "age": 30,
    "married": True,
    "divorced": False,
    "children": ("Ann","Billy"),
    "pets": None,
    "cars": [
        {"model": "BMW 230", "mpg": 27.5},
        {"model": "Ford Edge", "mpg": 24.1}
    ]
}

print(json.dumps(x))
```

```python
# use indentations and line breaks
json.dumps(x, indent=4)
```

```python
# use comma and space to separate each object
x = {
    "name": "John",
    "age": 30,
    "married": True,
    "divorced": False,
    "children": ("Ann","Billy"),
    "pets": None,
    "cars": [
        {"model": "BMW 230", "mpg": 27.5},
        {"model": "Ford Edge", "mpg": 24.1}
    ]
}
json.dumps(x, indent=4, separators=(".","="))
```

### RegEx Module

```python
import re

txt = "The rain in Spain"
x = re.search("^The.*Spain$",txt)
```

```python
# the findall() function returns a list containing all matches
txt = "The rain in Spain"
x = re.findall("ai",txt)
print(x)
```

```python
txt = "The rain in Spain"
x = re.findall("Portugal",txt)
print(x)
```

```python
# the search() function
import re

txt = "The rain in Spain"
x = re.search("Portugal",txt)
print(x)
```

```python
# split at each white-space character
import re

txt = "The rain in Spain"
x = re.split(r"\s",txt)
print(x)
```

```python
# a match object is an object containing information about the search and the result
import re

txt = "The rain in Spain"
x = re.search("ai",txt)
print(x)   # this will print an object
```

```python
# look for any words that starts with an upper case "S"
import re

txt = "The rain in Spain"
x = re.search(r"\bS\w+",txt)
print(x.span())
print(x.string)
print(x.group())
```

## Python file handling

```python
# read a text file
f = open("input_files/demofile.txt")
print(f.read())
```

```python
# using the with keyword
with open("input_files/demofile.txt") as f:
    print(f.read())
```

```python
# close the file once reading is finished
f = open("input_files/demofile.txt")
print(f.readline())
f.close()
```

```python
# loop through a file line by line
with open("input_files/demofile.txt") as f:
    for x in f:
        print(x)
```

```python
# write more content to an existing file
with open("input_files/demofile.txt","a") as f:
    f.write("Now the file has more content!")

# open and read the file after the appending
with open("input_files/demofile.txt") as f:
    print(f.read())
```
