```python
import sys, importlib

print("Python:", sys.version.split()[0])
print("Running from:", sys.executable, "\n")

packages = [
    "numpy", "pandas", "scipy", "matplotlib", "seaborn", "plotly", "sklearn",
    "torch", "torchvision", "tensorflow",
    "xgboost", "lightgbm", "catboost", "optuna", "shap", "statsmodels",
    "transformers", "datasets", "accelerate", "peft", "sentence_transformers", "diffusers",
    "langchain", "openai", "anthropic", "nltk", "spacy",
    "faiss", "chromadb", "cv2", "PIL", "timm", "ultralytics",
    "mlflow", "wandb", "polars", "pyarrow",
]

ok, missing = [], []
for p in packages:
    try:
        m = importlib.import_module(p)
        ok.append(f"{p:22} {getattr(m, '__version__', 'installed')}")
    except Exception as e:
        missing.append(f"{p:22} {type(e).__name__}: {str(e)[:60]}")

print("✅ Working:\n" + "\n".join(ok))
print("\n❌ Problems:\n" + ("\n".join(missing) if missing else "None 🎉"))
```

```python
import numpy as np, matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split

X, y = load_iris(return_X_y=True)
Xtr, Xte, ytr, yte = train_test_split(X, y, random_state=0)
print("Accuracy:", RandomForestClassifier().fit(Xtr, ytr).score(Xte, yte))

x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x)); plt.title("Plotting works"); plt.show()
```
